```python
from pyspark.sql.functions import col, year, month, weekofyear, to_date, dayofweek, dayofmonth, dayofyear, quarter
from pyspark.sql.types import DateType
from datetime import datetime, timedelta

# Input: start and end date strings in 'YYYY-MM-DD' format
start_date = '2025-01-01'
end_date = '2028-12-31'

# Generate date range
start_dt = datetime.strptime(start_date, '%Y-%m-%d')
end_dt = datetime.strptime(end_date, '%Y-%m-%d')
date_list = [(start_dt + timedelta(days=x)).date() for x in range((end_dt - start_dt).days + 1)]

# Create DataFrame
df = spark.createDataFrame([(d,) for d in date_list], ['date'])
df = df.withColumn('date', col('date').cast(DateType()))
df = df.withColumn('year', year(col('date'))) \
       .withColumn('month', month(col('date'))) \
       .withColumn('week', weekofyear(col('date'))) \
       .withColumn('day_of_week', dayofweek(col('date'))) \
       .withColumn('day_of_month', dayofmonth(col('date'))) \
       .withColumn('day_of_year', dayofyear(col('date'))) \
       .withColumn('quarter', quarter(col('date')))

# Write to retail_q.retail_gold.dim_calender table
df.write.format('delta').mode('overwrite').saveAsTable('retail_q.retail_gold.dim_calender')

display(df)
```
